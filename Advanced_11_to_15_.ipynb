```python
#Task:13
books = []

def add_book():
  book_id = input("Enter book ID: ")

  for book in books:
    if book["id"] == book_id:
      print("Book ID already exists!")
      return

  title = input("Enter book title: ")
  author = input("Enter author name: ")


  new_book = {
    "id": book_id,
    "title": title,
    "author": author,
    "available": True
}

  books.append(new_book)
  print(f"Book '{new_book['title']}' added!")

def display_books():
  if not books:
    print("No books available!")
    return

print("\n--- BOOKS AVAILABLE ---")
for book in books:
  status = "Available" if book["available"] else "Not Available"
  print(f"ID: {book['book_id']}, Title: {book['title']}, Author: {book['author']}, Status: {status}")


def search_book():

  search_title = input("Enter book title to search: ")
  found = False
  for book in books:
    if search_title in book ["title"].lower():
      status = "Available" if book["available"] else "Not Available"
      print(f"ID: {book['book_id']}, Title: {book['title']}, Author: {book['author']}, Status: {status}")
      found = True

  if not found:
    print("No books found!")

def borrow_book():
    book_id = input("Enter book ID to borrow: ")
    for book in books:
      if book["id"] == book_id:
        if book["available"]:
          book["available"] = False
          print(f"Book '{book['title']}' borrowed successfully!")
          return
        else:
          print("Book is not available for borrowing!")
          return
    print("Book not found!")

def return_book():
    book_id = input("Enter book ID to return: ")
    for book in books:
      if book["id"] == book_id:
        if not book["available"]:
          book["available"] = True
          print(f"Book '{book['title']}' returned successfully!")
        else:
          print("Book is already available!")
        return
      print("Book not found!")

while True:
  print("1. Add Book")
  print("2. Display Books")
  print("3. Search Book")
  print("4. Borrow Book")
  print("5. Return Book")
  print("6. Exit")

  choice = input("Enter choice: ")

  if choice == "1":
    add_book()
  elif choice == "2":
    display_books()
  elif choice == "3":
    search_book()
  elif choice == "4":
    borrow_book()
  elif choice == "5":
    return_book()
  elif choice == "6":
    print("Program ended!")
    break
  else:
    print("Invalid choice!")
```

```python
#Task:14
menu = {"Burger": 500, "Pizza": 1200, "Pasta": 800, "Drink": 200}

print("Menu:")
for item, price in menu.items():
  print(f"{item}: Rs. {price}")
print("----------------------------------")


order_list = []

while True:
  item_name = input("/n Enter item name (or 'done' to finish): ")

  if item_name.lower() == "done":
    break

  if item_name not in menu:
    print("Item not found in the menu!")
    continue

  quantity = int(input(f"Enter quantity for {item_name}: "))
  price = menu[item_name]
  subtotal = price * quantity

  order_list.append({
    "item": item_name,
    "quantity": quantity,
    "price": price,
    "subtotal": subtotal
  })
  print(f"{quantity} * {item_name} ")

total_bill = 0
for order in order_list:
  total_bill = total_bill + order["subtotal"]

tax = total_bill * 0.13

discount = 0
if total_bill >= 3000:
  discount = total_bill * 0.10

final_bill = total_bill + tax - discount


def show_receipt():
  print("\n========FINAL RECEIPT========")
  for order in order_list:
    i = order["item"]
    q = order["quantity"]
    p = order["price"]
    s = order["subtotal"]
    print(f"{i} ({q} x {p}) = Rs. {s}")

  print("----------------------------------")
  print(f"Total Bill: Rs. {total_bill}")
  print(f"Tax (13%): Rs. {tax}")
  if discount > 0:
    print(f"Discount (10%): -  Rs. {discount}")
  print(f"Final Payable Amount: Rs. {final_bill}")
  print("==================================")
  print("Thank you! Visit Again")

show_receipt()

```
